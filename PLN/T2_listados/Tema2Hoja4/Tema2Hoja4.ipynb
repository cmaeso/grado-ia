# Ejercicio 2

Se pide normalizar un texto a partir del uso de ER. Dado el siguiente texto, hay que normalizarlo haciendo varias tareas:
 - Eliminar signos de puntuación.
 - Eliminar palabras vacías del texto.
 - Eliminar todas las palabras con más de 5 caracteres.

```python
from nltk.corpus import stopwords
nltk.download('stopwords')
```

```python

text = """Francia aspira a jugar un papel protagonista en el auge de los
algoritmos. Esta semana, París fue el epicentro de una cumbre mundial sobre IA,
donde expertos de todo el mundo se adentraron en las amenazas y promesas de esta
tecnología. En el marco de este evento, el presidente Emmanuel Macron y el jeque
Mohamed bin Zayed Al Nahyan, líder de los Emiratos Árabes Unidos, presenciaron la
firma de un acuerdo de cooperación entre sus países, un pacto que promete
potenciar el desarrollo de proyectos conjuntos.
Como recoge la Agencia de Noticias de los Emiratos, la alianza incluye una inversión
por parte de la nación rica en petróleo en Francia, así como “la adquisición de chips
de vanguardia, la infraestructura de centros de datos y el desarrollo de talento, y
mediante el establecimiento de Embajadas de Datos Virtuales para permitir la IA
soberana y la infraestructura en la nube en ambos países”. El Gobierno francés, por
su parte, ha señalado que la iniciativa contempla la construcción de un enorme
centro de datos.
"""

no_punct= re.sub(r'[^\w\s]', ' ', text)
print(no_punct)
stop_words= set(stopwords.words('spanish'))
texto=""
no_punct_token= word_tokenize(no_punct, language='spanish')
print(no_punct_token, "\n")
for w in no_punct_token:
    if w.lower() not in stop_words:
         texto+= w +" "
print(texto, "\n")

less_5_c= re.sub(r'\b\w{6,}\b', '', texto) #eliminamos palabras con mas de 5 caracteres
less_5_c= re.sub(r'\s+', ' ', less_5_c) #quitamos espacios
less_5_c= less_5_c.strip()
 
print(less_5_c, "\n")

```

# Ejercicio 3

Dado el texto “Este es un texto que tiene ejemplos de fechas. Hoy es 09/02/2025, esta es una fecha posterior al 13 de enero de 2025. ¿Nos gustaría estar ya a 5 de julio y empezar las vacaciones? Casi que mejor no, que el tiempo avance a su ritmo. El primer día de clase fue el 30-01-2025, y el último día será el 8 de mayo del 2025.” Hay que construir expresiones regulares para que sean capaces de reconocer los diferentes formatos de fechas que aparecen (DD/MM/AAAA, DD-MM-AAAA, DD de MM de AAAA y DD de MM).

```python
texto= "Este es un texto que tiene ejemplos de fechas. Hoy es 09/02/2025, esta es una fecha posterior al 13 de enero de 2025. ¿Nos gustaría estar ya a 5 de julio y empezar las vacaciones? Casi que mejor no, que el tiempo avance a su ritmo. El primer día de clase fue el 30-01-2025, y el último día será el 8 de mayo del 2025."
```

```python
g= r"""
   (\d{1,2}(?:/|-)\d{1,2}(?:/|-)\d{4}
    |\d{1,2}\s+de\s+\w+\s*(?:del?\s+\d{4})?)
"""

print(re.findall(g, texto, re.VERBOSE)) #re.VERBOSE para ignorar saltos de línea y dem
```

# Ejercicio 4

Dado un texto que contenga números de teléfono de España, fijos (puede comenzar por el prefijo 8 o 9) o móviles (pueden comenzar con 6 o 7) y formados por 9 dígitos en total.

```python
text = "Hola soy x y mi móvil es 678903423 y mi fijo es 826738495"
```

```python
g= r"""
((?:6|7)\d{8} |
(?:9|8)\d{8})
"""
gram = r"[6-9]\d{8}"

print(re.findall(g, text, re.VERBOSE)) #re.VERBOSE para ignorar saltos de línea y dem
```
