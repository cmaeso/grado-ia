# Ejercicio 1


- Dado el texto siguiente, debemos tokenizarlo y:
    - Mostrar cada token en una línea diferente.
    - Añadir los diferentes tokens a una lista y mostrar la lista.
    - Mostrar para cada token y en una línea diferente, el texto del token, la categoría gramatical y el lema.
    - Contar el número de palabras vacías que se encuentran en el texto. Imprimir el número de palabras vacías y luego imprimir el número y las palabras vacías sin repeticiones.
    - Mostrar los adjetivos que aparecen en el texto.
    - Mostrar los nombres que aparecen en el texto. 
    - Mostrar para cada token su categoría gramatical y la explicación asociada a dicha categoría. Para ello se puede utilizar spacy.explain, que da una descripción acerca del tag particular del token.


```python
# Texto de ejemplo
text = "El 2025 viene cargado de innovaciones que repercutirán en nuestro modo de vida y de trabajo diario. Este año, las tendencias apuestan por los avances en inteligencia artificial (IA), automatización, conectividad y sostenibilidad. No sólo prometen una mayor eficiencia en el trabajo y una optimización de los procesos de decisión, sino también la superación de muchos de los retos actuales. Estas tendencias brindarán nuevas oportunidades tanto a empresas como a particulares. Echa un vistazo a las 5 tendencias tecnológicas más importantes, que marcarán el 2025."
# Cargamos el modelo
nlp = spacy.load("es_core_news_sm")
doc = nlp(text)
```

```python
# Cada token en una línea
for t in doc:
    print(f"Token: {t.text}")
          
```

```python
# Añadir los diferentes tokens a una lista y mostrar la lista.
list_tokens=[]
for i in doc:
    list_tokens.append(i.text)
print(list_tokens)
```

```python
#Mostrar para cada token y en una línea diferente, el texto del token, la categoría gramatical y el lema.
for i in doc:
    print(f"Token: {i.text}; categoría gramatical: {i.tag_}; lema: {i.lemma_}")
```

```python
#Número palabras vacías (stopwords)

stop_words=[]
for token in doc:
    if token.is_stop:
        stop_words.append(token.text)

print("Stopwords totales:", len(stop_words))
print("Sin repeticiones:" , len(set(stop_words)))

print(set(stop_words))
```

```python
#Mostrar adjetivos
adjetivos=[]

for token in doc:
    if token.tag_== 'ADJ':
        adjetivos.append(token.text)

print(adjetivos)
```

```python
# Mostrar sustantivos
sustantivos=[]
for token in doc:
    if token.tag_ == 'NOUN':
        sustantivos.append(token.text)

print(sustantivos)
    
```

```python
# Mostrar para cada token su categoría gramatical y la explicación asociada a dicha categoría.
#Para ello se puede utilizar spacy.explain, que da una descripción acerca del tag particular del token.
for token in doc:
    print(f"Token: {token.text} \nCategoría gramatical: {token.tag_};\nExplicación: {spacy.explain(token.tag_)}\n")
```

# Ejercicio 3

Dado el mismo texto en español de los ejercicios anteriores se quieren calcular estadísticas sobre frecuencias de las palabras del texto. Para ello, primero eliminar las palabras vacías y los signos de puntuación y sobre el listado de palabras resultante se pide:
-  Obtener la frecuencia de cada palabra, imprimiendo el par palabra-frecuencia.
- Mostrar sólo aquellas palabras cuya frecuencia sea mayor que 1.

```python
from collections import Counter

filtered_words=[]

for token in doc:
    if not token.is_stop and not token.is_punct:
        filtered_words.append(token.text.lower())

frecuency= Counter(filtered_words) 

for w, f in frecuency.items():
    print(w, f)

print("\n")

#imprimir solo frecuencias <1
for w, f in frecuency.items():
    if f>1:
        print(w, f)

```

# Ejercicio 4 

Implementar una función para preprocesar un texto con operaciones típicas para
normalizarlo, es decir, dado un texto se debe convertir a minúscula, lematizar cada
token, eliminar stop words y los signos de puntuación. (Lo que se pide es agrupar
estas operaciones en una única función para que se pueda invocar como una “única
acción” en un momento dado)

```python
def preprocesado(texto, nlp):
    
    doc= nlp(texto.lower())
    tokens_preprocesados=[]
    for t in doc:
        if not t.is_punct and not t.is_stop:
            tokens_preprocesados.append(t.lemma_)
    return " ".join(tokens_preprocesados)
```

# Ejercicio 5

Dados tres textos cortos en español, dos de la misma temática y uno de temática
diferente (openAustralia1.txt, openAustralia2.txt y crisisUcrania.txt), se pide:
- Comprobar cuál es su similitud (entre cada par de documentos), haciendo
uso del atributo similarity de la clase Doc y sin hacer ningún tipo de
preprocesamiento del texto.
- Comprobar cuál es su similitud una vez que se procesan los textos eliminando
signos de puntuación, espacios, palabras vacías y aquellos tokens con
longitud inferior a 4 caracteres. Una vez realizado este preprocesamiento, se
obtendría de nuevo un objeto de tipo Doc para cada uno y se comprobaría la
similitud.
- ¿Y si además del preprocesamiento anterior se lematizan los tokens? ¿Hay
variaciones en la similitud?

```python
def preprocesado_5b(texto, nlp):
    
    doc= nlp(texto.lower())
    tokens_preprocesados=[]
    for t in doc:
        if not t.is_punct and not t.is_stop and len(t.text) >=4 :
            tokens_preprocesados.append(t.lemma_)
    return " ".join(tokens_preprocesados)



rutas=["C:/Users/carme/Desktop/PLN/Tema2Hoja2/TextosListadoII/openAustralia1.txt",
       "C:/Users/carme/Desktop/PLN/Tema2Hoja2/TextosListadoII/openAustralia2.txt",
       "C:/Users/carme/Desktop/PLN/Tema2Hoja2/TextosListadoII/crisisUcrania.txt"]
textos=[]
for r in rutas:
    with open(r, "r", encoding="utf-8") as archivo:
        textos.append(archivo.read())
### requiere modelo más grande 
docs=[]
nlp= spacy.load('es_core_news_md')

### SIN PREPROCESAR ####
for t in textos:
    docs.append(nlp(t))
    
similitud_1_2=  docs[0].similarity(docs[1])
similitud_1_3= docs[0].similarity(docs[2])
similitud_2_3= docs[1].similarity(docs[2])

print("Sin preprocesar")
print(similitud_1_2, similitud_1_3 , similitud_2_3)

# PROCESADO B
docs_procesados=[]
for t in textos:
    text_preprocesado=preprocesado_5b(t,nlp)
    docs_procesados.append(nlp(text_preprocesado))

similitud_1_2=  docs_procesados[0].similarity(docs_procesados[1])
similitud_1_3= docs_procesados[0].similarity(docs_procesados[2])
similitud_2_3= docs_procesados[1].similarity(docs_procesados[2])

print("Preprocesado b ")
print(similitud_1_2, similitud_1_3 , similitud_2_3)
#PROCESADO + LEMATIZACION

docs_procesados=[]
for t in textos:
    text_preprocesado=preprocesado(t,nlp)
    docs_procesados.append(nlp(text_preprocesado))

similitud_1_2=  docs_procesados[0].similarity(docs_procesados[1])
similitud_1_3= docs_procesados[0].similarity(docs_procesados[2])
similitud_2_3= docs_procesados[1].similarity(docs_procesados[2])

print("Preprocesado + lematización ")
print(similitud_1_2, similitud_1_3 , similitud_2_3)
```

# Ejercicio 6

Dado el texto contenido en el archivo “atp.txt” se pide extraer la siguiente
información utilizando el matcher de spacy:
- Poner un patrón para obtener la cadena “(ATP)”
- Poner un patrón para obtener la cadena “Súper challengers”
-Poner un patrón para obtener las cadenas de texto con nombres y adjetivos.

Los atributos válidos para especificar en los patrones, así como los operadores de
cuantificación, entre otras cosas, se pueden encontrar aquí:
https://spacy.io/usage/rule-based-matching.

```python
import spacy
from spacy.matcher import Matcher

nlp =spacy.load("es_core_news_sm")
matcher = Matcher(nlp.vocab)  # Necesita el vocabulario del modelo

doc = nlp("El jugador español ganó el torneo de (ATP) en los Súper challengers.")
#Pattern con tokens consecutivos
pattern= [{"TEXT":"("},{"TEXT":"ATP"}, {"TEXT":")"}]
#Patterns inteligentes

pattern2 = [{"POS": "NOUN"}, {"POS": "ADJ"}]

matcher.add("ATP_PATTERN", [pattern])

matcher.add("Noun_adj_pattern", [pattern2])

matches = matcher(doc) 
# en matches se almacenan los resultados con el formato (id, inicio, fin)
for match_id, start, end in matches:
    pattern_name = nlp.vocab.strings[match_id]  # Obtiene el nombre del patrón
    span= doc[start:end] # Obtiene el fragmento del texto que hizo match

    print(f"Patrón: {pattern_name}, Match : {span.text}")
```
