# Ejercicio 4

Dado un fichero con contenido en inglés ("Data_Science.txt") se quieren eliminar todos los signos de puntuación que aparezcan en el mismo.

# Ejercicio 5

Dado un fichero de texto con contenido en inglés ("Data_Science.txt") se quiere saber cuáles son las 5 palabras más frecuentes del texto, antes y después de eliminar los signos de puntuación. Además, se quiere saber cuántas veces aparece la palabra "data" en el texto.
Se puede consultar la API de NLTK: https://www.nltk.org/api/nltk.probability.html

```python
ruta= "/mnt/c/Users/carme/Desktop/PLN/Tema2Hoja1/Textos/Data_Science.txt"
import string
puntuation= string.punctuation
from nltk.probability import FreqDist 
# FreqDist devuelve frecuencia de cada palabra, métodos útiles most_common(n) , freq(elemento)

with open(ruta, "r", encoding="utf-8") as archivo:
    texto_DS= archivo.read()
archivo.close()

DS_sentences= tokenizer.tokenize(texto_DS)
words_punct_filtered=[]
words=[]
for s in DS_sentences:
    token_words_filtered=[]
    token_words= word_tokenize(s)
    for w in token_words:
        words.append(w)
        if w not in puntuation:
            words_punct_filtered.append(w)

dist_punct_filtered= FreqDist(words_punct_filtered)
dist_non_filtered= FreqDist(words)

print("5 palabras más comunes sin filtrado:", dist_non_filtered.most_common(5))
print("5 palabras más comunes con filtrado:", dist_punct_filtered.most_common(5))

```

obtener su contenido, dividirlo en frases y por cada una de ellas eliminar las palabras
vacías (stop words) que contengan. 

```python
from nltk.corpus import stopwords
nltk.download('stopwords')

with open(ruta, "r", encoding="utf-8") as archivo:
    texto_DS=archivo.read()
archivo.close()

stop_words= set(stopwords.words('english'))

DS_sentences= tokenizer.tokenize(texto_DS)


for s in DS_sentences:
    words_tokenized= word_tokenize(s)
    no_stopwords=[]

    for w in words_tokenized:
        if w.lower() not in stop_words and w not in puntuation:
            no_stopwords.append(w)
            
    print(no_stopwords)    
```

# Ejercicio 12

Dado un fichero de texto con contenido en inglés (“Data_Science.txt”) se pide obtener los diferentes tokens del texto y por cada uno de ellos lematizar.
