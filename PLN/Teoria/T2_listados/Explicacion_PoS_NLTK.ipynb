```python
# Con la librería NLTK se puede hacer part-of-speech (POS) para inglés. Tiene soporte para otros idiomas dando la opción 
# de entrenar taggers propios. 

# Ejemplo de pos por defecto 
from nltk import pos_tag, word_tokenize, pos_tag_sents
    
# No haría falta añadir el idioma al final, porque por defecto lo hace en inglés    
print(pos_tag(word_tokenize("John's big idea isn't all that bad."),lang="eng"))

print("--------------------")

print(pos_tag(word_tokenize("John's big idea isn't all that bad.")))

```

```python
# ¿Qué ocurre si queremos hacer pos en español?

print(pos_tag(word_tokenize("El hombre bajo tocaba el bajo bajo la escalera.")))

# Obtenemos una salida, pero no es correcta, porque no está etiquetando correctamente cada token

# ¿Y si especificamos el idioma? Siguiendo la nomenclatura de códigos ISO 639 para especificar el idioma, que es como lo 
# requiere NLTK
print(pos_tag(word_tokenize("El hombre bajo tocaba el bajo bajo la escalera."), lang="es"))

# Nos da error, nos dice que " "Currently, NLTK pos_tag only supports English and Russian (i.e. lang='eng' or lang='rus')""
```

```python
# Podemos utilizar spaCy para hacer el pos en español
import spacy

# se carga el modelo
nlp = spacy.load('es_core_news_sm')

texto = "El hombre bajo tocaba el bajo bajo la escalera."

doc = nlp(texto)
 
# Borrar stop words y signos de puntuación
for token in doc: 
    print(token.text, ' -- ', token.pos_)
```
