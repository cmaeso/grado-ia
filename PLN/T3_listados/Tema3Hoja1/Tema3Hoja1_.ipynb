# Ejercicio 1
Utilizando una representación de bolsa de palabras (BoW) se quiere que los rasgos del
vocabulario no sean palabras, sino n-gramas. Los n-gramas son secuencias contiguas de n elementos en un texto, cruciales para captar la información contextual. Pueden mejorar la representación de los textos al no considerar palabras sueltas, sino combinaciones de estas, lo que proporciona un contexto más rico.  
Dadas las siguientes frases: “Estamos ya a finales de febrero.”, “En febrero sigue haciendo frío.”, “Esto es una frase de ejemplo que habla de un mes del año.” Se pide lo siguiente:
- Generar una representación BoW con bigramas (2-grams).
- Generar una representación BoW con unigramas, bigramas y trigramas.
En cada caso se debe de imprimir el vocabulario generado y la matriz de rasgosdocumentos.

## Bigramas

```python
corpus = [
    "Estamos ya a finales de febrero.",
    "En febrero sigue haciendo frío.",
    "Esto es una frase de ejemplo que habla de un mes del año."
]

vectorizer = CountVectorizer(ngram_range=(2,2))

X = vectorizer.fit_transform(corpus)
print("Matriz de características (vectorizer):")
print(X.toarray())
print("vectorizer (vectorizer):")
print(vectorizer.get_feature_names_out())

```

## Unigramas, bigramas y trigramas

```python
corpus = [
    "Estamos ya a finales de febrero.",
    "En febrero sigue haciendo frío.",
    "Esto es una frase de ejemplo que habla de un mes del año."
]

vectorizer_uni_to_tri= CountVectorizer(ngram_range=(1,3))
X_uni_to_tri = vectorizer_uni_to_tri.fit_transform(corpus)
print("Matriz de características (vectorizer_uni_to_tri):")
print(X_uni_to_tri.toarray())
print("vectorizer_uni_to_tri (vectorizer_uni_to_tri):")
print(vectorizer_uni_to_tri.get_feature_names_out())
```

```python

# 1. Cargar documentos y etiquetas
docs = []
labels = []
for cat in movie_reviews.categories():
    for fileid in movie_reviews.fileids(cat):
        text = movie_reviews.raw(fileid)
        docs.append(text)
        labels.append(cat)

#Crear un DataFrame
df = pd.DataFrame({'text': docs, 'label': labels})
X= df['text']
y= df['label']


# Cargar el modelo de spaCy
nlp = spacy.load("es_core_news_sm")  

# 2. Preprocesamiento estándar
def preprocess_spacy(docs, remove_stopwords=True, lemmatize=True):
    preprocessed_docs = []

    for doc_text in docs:
        doc = nlp(doc_text)
        tokens = []

        for token in doc:
            if not token.is_alpha:
                continue
            if remove_stopwords and token.is_stop:
                continue
            if lemmatize:
                tokens.append(token.lemma_.lower())
            else:
                tokens.append(token.text.lower())

        preprocessed_docs.append(" ".join(tokens))

    return preprocessed_docs

docs_preprocessed = preprocess_spacy(X, remove_stopwords=True, lemmatize=True)
```

```python
#Dividir el conjunto de datos en entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(docs_preprocessed, y, test_size=0.2, random_state=42)

# 3. Vectorización con CountVectorizer
vect = CountVectorizer(binary=True)
bow_b = vect.fit_transform(X_train)
features_b = vect.get_feature_names_out()
matrix_b = bow_b.toarray()  

# 4. Vectorización con CountVectorizer (sin binarizar)
vect_tf = CountVectorizer(binary=False)
bow_tf = vect_tf.fit_transform(X_train)
matrix_tf = bow_tf.toarray()  

#5 . Vectorización con TF-IDF
vectorizer_tfidf = TfidfVectorizer()
bow_tfidf = vectorizer_tfidf.fit_transform(X_train)
matrix_tfidf = bow_tfidf.toarray()

# 6. Entrenamiento del modelo Naive Bayes
model1 = MultinomialNB()
model2 = MultinomialNB()
model3 = MultinomialNB()

model1.fit(bow_b, y_train)
model2.fit(bow_tfidf, y_train)
model3.fit(bow_tf, y_train)


# 7. Predicción y evaluación
y_pred_b = model1.predict(vect.transform(X_test))
y_pred_tfidf = model2.predict(vectorizer_tfidf.transform(X_test))
y_pred_tf = model3.predict(vect_tf.transform(X_test))

# 8. Evaluación de precisión
accuracy_b = accuracy_score(y_test, y_pred_b)
accuracy_tfidf = accuracy_score(y_test, y_pred_tfidf)       
accuracy_tf = accuracy_score(y_test, y_pred_tf)

print("Precisión con CountVectorizer (binary=True):", accuracy_b)
print("Precisión con CountVectorizer (binary=False):", accuracy_tf)
print("Precisión con TF-IDF:", accuracy_tfidf)

```
