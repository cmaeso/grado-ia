```python
# Vectorizar "Headline"
nlp= spacy.load('es_core_news_md')
def preprocesado(text):
    docs= nlp(text)
    lemmas=''
    for t in docs:

        if t.is_alpha and t.pos_ in ['NOUN', 'ADJ', 'VERB', 'ADV', 'PRON']:
            lemmas+=t.lemma_+' '
    return lemmas.strip()

print(df['Headline'].head())

df['Headline']= df['Headline'].apply(preprocesado) 
print(df['Headline'].head())

vect= TfidfVectorizer()
bow= vect.fit_transform(df['Headline'])
matrix= bow.toarray()
features= vect.get_feature_names_out() 
df2=pd.DataFrame(matrix, columns=features)
print(df2)
```

```python
# calcular matrix similitud conseno
cosine_sim = cosine_similarity(bow)
sim_indices= cosine_sim[1].argsort()[-5:][::-1]
print(sim_indices)

print(cosine_sim[sim_indices])
```
